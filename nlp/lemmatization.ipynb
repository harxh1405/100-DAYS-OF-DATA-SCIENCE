## We saw how stemming gives unwanted results

## Output of stemming is root stem

## Output of lemmatization is root word(does not change the meaning)

```python
from nltk.stem import WordNetLemmatizer
import nltk
nltk.download('wordnet')
wnl=WordNetLemmatizer()
wnl.lemmatize('cats') ##lemmatization is better than stemming as it does not destroy the meaning of the word and gives the root word as output
```

```python
'''
noun-n
verb-v
adjective-a
adverb-r
'''
wnl.lemmatize('better',pos='a') ##lemmatization is better than stemming as it does not destroy the meaning of the word and gives the root word as output
```

```python
words=['eating','eats','eaten','eat','write','writing','wrote','written','write','go','going','went','gone','program','programming','programmer','programmed','history','historical','historian','historically']
```

```python
for word in words:
    print(word+"---->"+wnl.lemmatize(word,pos='v')) ##lemmatization is better than stemming as it does not destroy the meaning of the word and gives the root word as output
```
