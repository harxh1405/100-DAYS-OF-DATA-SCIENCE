Stemming

It is the process of reducing a word to its word stem. E.g.

Classification - Positive/Negative review

Reviews---->eating,eat,eaten--->eat(root/lemma)
going,gone,goes--->go(root/lemma)

Each word represents a vector and this may increase the input size



```python
words=['eating','eats','eaten','eat','write','writing','wrote','written','write','go','going','went','gone','program','programming','programmer','programmed','history','historical','historian','historically']
```

```python
### Porter Stemmer
from nltk.stem import PorterStemmer
ps=PorterStemmer()
```

```python
for word in words:
    print(word+"---->"+ps.stem(word))

    
```

```python
ps.stem('history') ##for some of the words it may give ambiguous results totally destroying the meaning of the word
```

```python
##RegexpStemmer class
from nltk.stem import RegexpStemmer
```

```python
reg_stemmer=RegexpStemmer('ing$|s$|e$|able$',min=4) ##$->last
reg_stemmer.stem('eating')
```

```python
##Snowball Stemmer(better)

from nltk.stem import SnowballStemmer
snow_stemmer=SnowballStemmer('english')
for word in words:
    print(word+"---->"+snow_stemmer.stem(word))
```

```python
ps.stem('fairly'),ps.stem('sportingly')
```

```python
snow_stemmer.stem('fairly'),snow_stemmer.stem('sportingly')
##Snowball stemmer performs better here
```
