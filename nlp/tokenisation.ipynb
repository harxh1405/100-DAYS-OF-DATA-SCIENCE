```python
corpus="""Hello welcome, I am Harsh Singh. I am a data scientist. I have 3 years of experience in the field of data science. I have worked on various projects related to data science. I have a good knowledge of machine learning, deep learning, and natural language processing. I am also proficient in Python programming language."""
```

```python
corpus
```

```python
###Tokenisation
###sentence ---> paragraph
from nltk.tokenize import sent_tokenize
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
documents=sent_tokenize(corpus)
```

```python
documents
```

```python
###Tokenisation

##paragraph-->words
##sentence-->words
from nltk.tokenize import word_tokenize
word_tokenize(corpus)
```

```python
for sentence in documents:
    print(word_tokenize(sentence))
```

```python
from nltk.tokenize import wordpunct_tokenize
wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordTokenizer
```

```python
tokenizer=TreebankWordTokenizer()
```

```python
tokenizer.tokenize(corpus)
```
