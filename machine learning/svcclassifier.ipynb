```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
###Lets create synthetic data points
from sklearn.datasets import make_classification
```

```python
X,y=make_classification(n_samples=1000, n_features=2,n_classes=2, n_clusters_per_class=2, random_state=4, n_redundant=0) 
###play with different n_clusters_per_class values to see the effect on data separability
##for example n_clusters_per_class=1 will make the data linearly separable
##for n_clusters_per_class=2 or more the data will be non-linearly separable
X
```

```python
X=pd.DataFrame(X,columns=['feature1','feature2'])
```

```python
sns.scatterplot(x=X.iloc[:,0],y=X.iloc[:,1],hue=y)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

```python
from sklearn.svm import SVC
```

```python
svc=SVC(kernel='linear')##since the data is linearly separable
svc.fit(X_train,y_train)
```

```python
y_pred=svc.predict(X_test) 
```

```python
from sklearn.metrics import confusion_matrix, classification_report,accuracy_score
```

```python
print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
print("Accuracy:",accuracy_score(y_test,y_pred))
```

```python
rbf_svc=SVC(kernel='rbf')##using rbf kernel
rbf_svc.fit(X_train,y_train)
```

```python
y_rbf_pred=rbf_svc.predict(X_test)
```

```python
print(confusion_matrix(y_test,y_rbf_pred))
print(classification_report(y_test,y_rbf_pred))
print("Accuracy:",accuracy_score(y_test,y_rbf_pred)) ###accuracy will be improved using rbf kernel
```

```python
poly_svc=SVC(kernel='poly',degree=3)##using polynomial kernel
poly_svc.fit(X_train,y_train)
```

```python
y_pred_poly=poly_svc.predict(X_test)
```

```python
print(confusion_matrix(y_test,y_pred_poly))
print(classification_report(y_test,y_pred_poly))
print("Accuracy:",accuracy_score(y_test,y_pred_poly))
```

```python
sigmoid_svc=SVC(kernel='sigmoid')##using sigmoid kernel
sigmoid_svc.fit(X_train,y_train)
```

```python
y_pred_sigmoid=sigmoid_svc.predict(X_test)
```

```python
print(confusion_matrix(y_test,y_pred_sigmoid))
print(classification_report(y_test,y_pred_sigmoid))
print("Accuracy:",accuracy_score(y_test,y_pred_sigmoid))
```

###Hyperparameter tuning

```python
from sklearn.model_selection import GridSearchCV
param_grid={'C':[0.1,1,10,100],'gamma':[1,0.1,0.01,0.001],'kernel':['rbf','poly','sigmoid']}

```

```python
grid_search=GridSearchCV(SVC(),param_grid,refit=True,cv=5,verbose=3)
```

```python
grid_search.fit(X_train,y_train)
```

```python
grid_search.best_params_
```

```python
y_pred_best=grid_search.predict(X_test)
```

```python
print(confusion_matrix(y_test,y_pred_best))
print(classification_report(y_test,y_pred_best))
print("Accuracy:",accuracy_score(y_test,y_pred_best))
```
