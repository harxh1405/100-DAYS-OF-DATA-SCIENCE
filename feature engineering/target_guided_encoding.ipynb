```python
import pandas as pd
df=pd.DataFrame({
    'city':['New York','London','Paris','Tokyo','New York','Paris'],
    'price':[200,150,300,250,100,320]
})
```

```python
df
```

```python
mean_price=df.groupby('city')['price'].mean().to_dict()
mean_price
```

```python
df['city_encoded']=df['city'].map(mean_price)
```

```python
df
```

```python
new_df=df[['city','city_encoded']]
new_df
```
